```python
path = "C:\\Users\\alvin\\OneDrive\\Documents\\CS171\\project\\breast-cancer-classifier-cs171-s02"
benign_full = path + "\\data\\benign\\"
malignant_full = path + "\\data\\malignant"

benign_images = os.listdir(benign_full)
malignant_images = os.listdir(malignant_full)

# Sampling 30% of both benign and malignant datasets
n_sample = int(0.3 * len(benign_images))

benign_files = [f for f in os.listdir(benign_full) if f.endswith(('.jpg', '.jpeg', '.png'))]
malignant_files = [f for f in os.listdir(malignant_full) if f.endswith(('.jpg', '.jpeg', '.png'))]

indices_benign = np.random.choice(len(benign_files), size=n_sample, replace=False)
indices_malignant = np.random.choice(len(malignant_files), size=n_sample, replace=False)

benign_samples_dir = "samples/benign"
malignant_samples_dir = "samples/malignant"

os.makedirs(benign_samples_dir, exist_ok=True)
os.makedirs(malignant_samples_dir, exist_ok=True)

# Add benign samples to the 'samples/' directory
for idx in indices_benign:
    img_path = os.path.join(benign_full, benign_files[idx])
    shutil.copy(img_path, benign_samples_dir)

# Add malignant samples to the 'samples/' directory
for idx in indices_malignant:
    img_path = os.path.join(malignant_full, malignant_files[idx])
    shutil.copy(img_path, malignant_samples_dir)
```
