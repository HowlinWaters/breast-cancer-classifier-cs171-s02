```python
from pathlib import Path
import numpy as np
import shutil
```

```python
path = Path().resolve().parent
benign_full = path / "data" / "benign"
malignant_full = path / "data" / "malignant"

benign_images = list(benign_full.iterdir())
malignant_images = list(malignant_full.iterdir())

# Sampling 30% of both benign and malignant datasets
n_sample = int(0.3 * len(benign_images))

benign_files = [f.name for f in benign_full.iterdir()]
malignant_files = [f.name for f in malignant_full.iterdir()]

indices_benign = np.random.choice(len(benign_files), size=n_sample, replace=False)
indices_malignant = np.random.choice(len(malignant_files), size=n_sample, replace=False)

benign_samples_dir = path / "samples" / "benign"
malignant_samples_dir = path / "samples" / "malignant"

benign_samples_dir.mkdir(parents=True, exist_ok=True)
malignant_samples_dir.mkdir(parents=True, exist_ok=True)

# Add benign samples to the 'samples/' directory
for idx in indices_benign:
    img_path = benign_full / benign_files[idx]
    shutil.copy(img_path, benign_samples_dir)

# Add malignant samples to the 'samples/' directory
for idx in indices_malignant:
    img_path = malignant_full / malignant_files[idx]
    shutil.copy(img_path, malignant_samples_dir)
```
